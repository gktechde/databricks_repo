```python

sample_df= (spark.read\
            .option("header","true")\
            .option("sep",'\t')\
            .option("infer_schema","true")\
            .csv("/Volumes/exp_catalog/exp_schema/exp_volume/")\
)

exp_sch = sample_df.schema

spark.readStream.format("csv")\
    .option("header",'true')\
    .option("sep",'\t')\
    .schema(exp_sch)\
    .option("infer_schema",'true')\
    .load("/Volumes/exp_catalog/exp_schema/exp_volume/")\
    .writeStream\
    .format("delta")\
    .option("checkpointLocation","/Volumes/exp_catalog/exp_schema/exp_volume/")\
    .trigger(availableNow=True)\
    .table("expense_bronze")

```
