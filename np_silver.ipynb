```python
from pyspark.sql.functions import col, trim, to_date

df = spark.readStream \
          .table("expense_bronze") 

df_silver = df \
            .drop("Note")\
            .drop("Account")\
            .withColumn("Category", trim(col("Category")))\
            .withColumn("expense_dt",to_date(trim(col("exp_date")),'MM/dd/yyyy'))

query =  df_silver.writeStream \
                  .format("delta")\
                  .option("checkpointLocation","/Volumes/exp_catalog/exp_schema/exp_volume/exp_silver")\
                  .trigger(availableNow=True)\
                  .outputMode("append")  \
                  .toTable("expense_silver")
  

```
